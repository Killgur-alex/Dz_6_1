### Задание *1*

Переведите содержимое файла purchase_log.txt в словарь purchases вида:{‘1840e0b9d4’: ‘Продукты’, …}


Пример работы программы при выводе первых двух элементов словаря purchases:

1840e0b9d4 ‘Продукты‘

4e4f90fcfb  ‘Электроника‘

```python
purchases = {}
```

```python
file = open("purchase_log.txt", "r", encoding="utf-8")
```

```python
next(file) # пропустили первую строку заголовком, так как она не нужна в решении задачи конкретно в данному случае
```

```python

for line in file:
  line = line.strip().strip('{}')
  if line:
      small_line  = line.split(',')

      user_id = ""
      category = ""

      for item in small_line:
            key, value = item.split(":")
            key = key.strip().strip('"')
            value = value.strip().strip('"')
# Из-за того что в файле все записано в тип строка, а ключ и значение в строке первой перепутаны, приходится проверять вручную соотвествие юзер айди и категории в каждой строке. для этого вложенный цикл.
#  при работе со словарем проблемы бы такой не было, мы бы обращались по ключу и не важно в каком порядке бы они шли.
            if key == "user_id":
                user_id = value
            elif key == "category":
                category = value


      
      purchases[user_id] = category

```

```python
file.close()
```

```python

count = 0
for user_id, category in purchases.items():
    print(user_id, f"'{category}'")
    count += 1
    if count == 5:
        break

```
